```python
import csv
import re

def parse_hgvs_cds(name_string):
    """
    Matches simple coding SNVs like c.1032G>A.
    Returns None for anything else (splice offsets, indels, etc.) — intentional,
    since this project is scoped to missense/coding-SNV only.
    """
    match = re.search(r'c\.(\d+)([ACGT])>([ACGT])', name_string)
    if match:
        pos, ref, alt = match.groups()
        return int(pos), ref, alt
    return None

GENE_COL = "Gene(s)"
LABEL_COL = "Germline classification"
KEEP_LABELS = {"Pathogenic", "Benign", "Uncertain significance"}
MAX_GENES_LISTED = 3  # excludes large structural variants spanning many genes

variants = []
dropped = {"wrong_gene": 0, "bad_label": 0, "unparseable": 0}

with open("clinvar_download_20260903_120920.tsv") as f:
    reader = csv.DictReader(f, delimiter='\t')
    for row in reader:
        gene_field = row.get(GENE_COL, "").strip()
        gene_list = gene_field.split("|")
        if "KCNQ1" not in gene_list or len(gene_list) > MAX_GENES_LISTED:
            dropped["wrong_gene"] += 1
            continue

        label = row.get(LABEL_COL, "").strip()
        if label not in KEEP_LABELS:
            dropped["bad_label"] += 1
            continue

        parsed = parse_hgvs_cds(row["Name"])
        if parsed is None:
            dropped["unparseable"] += 1
            continue

        cds_pos, ref, alt = parsed
        variants.append({
            "name": row["Name"],
            "cds_pos": cds_pos,
            "ref": ref,
            "alt": alt,
            "clinvar_label": label,
        })

print(f"Kept {len(variants)} variants")
print(f"Dropped: {dropped}")
```

```python
def is_missense(name):
    """
    True only for missense variants — excludes synonymous (p.===) 
    and stop-gained/nonsense (p.***Ter).
    """
    match = re.search(r'p\.([A-Za-z]{3})\d+([A-Za-z]{3}|=)', name)
    if not match:
        return True  # no protein annotation found — let it through, check separately
    ref_aa, alt_aa = match.groups()
    if alt_aa == '=' or alt_aa == 'Ter':
        return False
    return True

# rebuild missense-only pools correctly
benign_strict_missense = [v for v in by_label["Benign_strict"] if is_missense(v["name"])]
benign_likely_missense = [v for v in by_label["Benign_likely"] if is_missense(v["name"])]
pathogenic_missense = [v for v in by_label["Pathogenic"] if is_missense(v["name"])]
vus_missense = [v for v in by_label["VUS"] if is_missense(v["name"])]

print(f"Benign strict missense: {len(benign_strict_missense)}")
print(f"Benign likely missense: {len(benign_likely_missense)}")
print(f"Pathogenic missense available: {len(pathogenic_missense)}")
print(f"VUS missense available: {len(vus_missense)}")
```

```python
cadd_by_variant = {}

for line in lines:
    if line.startswith("#"):
        continue
    fields = line.strip().split("\t")
    row = dict(zip(columns, fields))
    key = (row["#Chrom"], row["Pos"], row["Ref"], row["Alt"])
    cadd_by_variant[key] = {
        "cadd_raw": row["RawScore"],
        "cadd_phred": row["PHRED"],
    }

print(f"Parsed CADD scores for {len(cadd_by_variant)} variants")

# merge into working_set using the genomic coords we pulled earlier from VEP
missing = []
for v in working_set:
    vep_key_name = re.sub(r'\s*\(p\..*?\)\s*$', '', v["name"])
    if vep_key_name in vep_genomic:
        g = vep_genomic[vep_key_name]
        cadd_key = (g["chrom"], g["pos"], g["ref"], g["alt"])
        if cadd_key in cadd_by_variant:
            v["cadd_phred"] = cadd_by_variant[cadd_key]["cadd_phred"]
        else:
            missing.append(v["name"])
    else:
        missing.append(v["name"])

print(f"Merged CADD scores: {len(working_set) - len(missing)} / {len(working_set)}")
if missing:
    print("Missing:", missing)
```

```python
missing = []
for v in working_set:
    vep_key_name = re.sub(r'\s*\(p\..*?\)\s*$', '', v["name"])
    if vep_key_name in vep_genomic:
        g = vep_genomic[vep_key_name]
        cadd_key = (g["chrom"], g["pos"], g["ref"], g["alt"])
        if cadd_key in cadd_by_variant_raw:  # see note below
            row = cadd_by_variant_raw[cadd_key]
            spliceai_vals = []
            for col in ["SpliceAI-acc-gain", "SpliceAI-acc-loss", "SpliceAI-don-gain", "SpliceAI-don-loss"]:
                val = row.get(col, "-")
                if val not in ("-", "", "NA"):
                    spliceai_vals.append(float(val))
            v["spliceai_max"] = max(spliceai_vals) if spliceai_vals else None
        else:
            missing.append(v["name"])

print(f"Merged SpliceAI scores: {sum(1 for v in working_set if v.get('spliceai_max') is not None)} / {len(working_set)}")
```

```python
key_found_but_no_score = 0
key_not_found_at_all = 0

for v in working_set:
    if v.get("spliceai_max") is not None:
        continue
    vep_key_name = re.sub(r'\s*\(p\..*?\)\s*$', '', v["name"])
    g = vep_genomic.get(vep_key_name)
    if g is None:
        key_not_found_at_all += 1
        continue
    cadd_key = (g["chrom"], g["pos"], g["ref"], g["alt"])
    if cadd_key not in cadd_by_variant_raw:
        key_not_found_at_all += 1
    else:
        row = cadd_by_variant_raw[cadd_key]
        raw_vals = [row.get(c, "?") for c in ["SpliceAI-acc-gain", "SpliceAI-acc-loss", "SpliceAI-don-gain", "SpliceAI-don-loss"]]
        print(v["name"], "->", raw_vals)
        key_found_but_no_score += 1

print(f"\nKey matched but SpliceAI columns empty/dash: {key_found_but_no_score}")
print(f"Key not found at all (real merge failure): {key_not_found_at_all}")
```

```python
def parse_score(s):
    m = re.search(r'\(([\d.]+)\)', s or "")
    return float(m.group(1)) if m else None

for v in working_set:
    v["sift_val"] = parse_score(v.get("sift"))
    v["polyphen_val"] = parse_score(v.get("polyphen"))
    v["cadd_phred_val"] = float(v["cadd_phred"])
    v["spliceai_reported"] = v.get("spliceai_max") is not None
    v["spliceai_val"] = v["spliceai_max"] if v["spliceai_reported"] else 0.0

metrics = ["cadd_phred_val", "sift_val", "polyphen_val", "spliceai_val"]
for label in ["Pathogenic", "Uncertain significance", "Likely benign", "Benign"]:
    group = [v for v in working_set if v["clinvar_label"] == label]
    parts = []
    for m in metrics:
        vals = [v[m] for v in group if v[m] is not None]
        mean = sum(vals) / len(vals) if vals else float("nan")
        parts.append(f"{m}={mean:.2f} (n={len(vals)})")
    print(f"{label}: " + " | ".join(parts))
```

```python
from Bio import AlignIO
import numpy as np

alignment = AlignIO.read("kcnq1_alligned.fasta", "fasta")
print([r.id for r in alignment])
print("Alignment length:", alignment.get_alignment_length())

def column_conservation(column, min_seqs=4):
    bases = [b.upper() for b in column if b != '-']
    if len(bases) < min_seqs:
        return None  # too few species to say anything
    counts = {}
    for b in bases:
        counts[b] = counts.get(b, 0) + 1
    total = len(bases)
    entropy = -sum((c/total) * np.log2(c/total) for c in counts.values())
    return (2 - entropy) / 2  # 1 = fully conserved, 0 = max variable

conservation = [column_conservation(alignment[:, i]) for i in range(alignment.get_alignment_length())]

# map human CDS position -> alignment column
human_row = next(str(r.seq) for r in alignment if "NM_000218.3" in r.id)
position_map = {}
cds_pos = 0
for col_idx, base in enumerate(human_row):
    if base != '-':
        cds_pos += 1
        position_map[cds_pos] = col_idx

# CHECK 1: does the human sequence start at the ATG?
print("First 3 human bases:", human_row.replace('-', '')[:3], "(should be ATG)")
print("Human ungapped length:", len(position_map))

# CHECK 2: does the human base at each mapped position match the ClinVar ref allele?
mismatches = []
for v in working_set:
    col = position_map.get(v["cds_pos"])
    if col is None:
        mismatches.append((v["name"], "position not in map"))
        continue
    if human_row[col].upper() != v["ref"]:
        mismatches.append((v["name"], f"alignment has {human_row[col].upper()}, ClinVar ref is {v['ref']}"))

print(f"Ref-allele mismatches: {len(mismatches)} / {len(working_set)}")
for m in mismatches[:5]:
    print(m)
```

```python
for v in working_set:
    col = position_map[v["cds_pos"]]
    v["conservation"] = conservation[col]
    v["n_species_at_site"] = sum(1 for r in alignment if r.seq[col] != '-')

# how many variants got no score because too few species aligned there?
none_count = sum(1 for v in working_set if v["conservation"] is None)
print(f"Variants with no conservation score: {none_count} / {len(working_set)}")

for label in ["Pathogenic", "Uncertain significance", "Likely benign", "Benign"]:
    vals = [v["conservation"] for v in working_set
            if v["clinvar_label"] == label and v["conservation"] is not None]
    if vals:
        print(f"{label}: n={len(vals)}, mean={sum(vals)/len(vals):.3f}, min={min(vals):.3f}, max={max(vals):.3f}")

# how conserved is the average site overall? (baseline for interpreting the above)
all_vals = [c for c in conservation if c is not None]
print(f"All scored alignment columns: mean={sum(all_vals)/len(all_vals):.3f}")
```

```python
import numpy as np
from sklearn.metrics import roc_auc_score

rng = np.random.default_rng(42)

# sign=-1 for SIFT because low score = damaging
methods = {"cadd_phred_val": 1, "sift_val": -1, "polyphen_val": 1,
           "spliceai_val": 1, "conservation": 1}

def auc_with_ci(pos, neg, key, sign, n_boot=2000):
    p = np.array([v[key] for v in pos]) * sign
    n = np.array([v[key] for v in neg]) * sign
    y = np.r_[np.ones(len(p)), np.zeros(len(n))]
    s = np.r_[p, n]
    auc = roc_auc_score(y, s)
    boots = []
    for _ in range(n_boot):
        pb = rng.choice(p, len(p), replace=True)
        nb = rng.choice(n, len(n), replace=True)
        boots.append(roc_auc_score(np.r_[np.ones(len(pb)), np.zeros(len(nb))], np.r_[pb, nb]))
    return auc, np.percentile(boots, 2.5), np.percentile(boots, 97.5)

path = [v for v in working_set if v["clinvar_label"] == "Pathogenic"]
sets = {
    "strict benign (n=4)": [v for v in working_set if v["clinvar_label"] == "Benign"],
    "expanded benign (n=12)": [v for v in working_set if v["clinvar_label"] in ("Benign", "Likely benign")],
}

for set_name, neg in sets.items():
    print(f"\n{set_name} vs Pathogenic (n={len(path)})")
    for key, sign in methods.items():
        a, lo, hi = auc_with_ci(path, neg, key, sign)
        print(f"  {key:16s} AUC={a:.3f}  95% CI [{lo:.3f}, {hi:.3f}]")
```

```python
def calls(v):
    return {
        "cadd": v["cadd_phred_val"] >= 20,
        "sift": v["sift_val"] is not None and v["sift_val"] < 0.05,
        "polyphen": (v.get("polyphen") or "").startswith("probably_damaging"),
        "conservation": v["conservation"] is not None and v["conservation"] >= 0.9,
    }

for v in working_set:
    c = calls(v)
    v["n_damaging"] = sum(c.values())
    v["votes"] = "".join(k[0].upper() if val else "-" for k, val in c.items())  # C S P C

# 1) how does the vote behave on the controls?
print("Votes (0-4 predictors saying damaging) by label:")
for label in ["Pathogenic", "Uncertain significance", "Likely benign", "Benign"]:
    cnt = Counter(v["n_damaging"] for v in working_set if v["clinvar_label"] == label)
    print(f"  {label:24s}", dict(sorted(cnt.items())))

# 2) the VUS table, most-split cases first
vus = [v for v in working_set if v["clinvar_label"] == "Uncertain significance"]
vus.sort(key=lambda v: abs(v["n_damaging"] - 2))
print("\nVUS (split cases first)  votes = Cadd Sift Polyphen Conservation")
for v in vus:
    print(f'{v["name"].split(":")[1]:34s} n={v["n_damaging"]} {v["votes"]}  '
          f'CADD={v["cadd_phred_val"]:5.1f} SIFT={v["sift_val"]:.2f} PP={v["polyphen_val"]:.2f} cons={v["conservation"]:.2f}')

pd.DataFrame(working_set)[cols + ["n_damaging", "votes"]].to_csv("kcnq1_classical_scores.csv", index=False)
```

```python
def calls(v):
    pp = v.get("polyphen") or ""
    return {
        "cadd": v["cadd_phred_val"] >= 20,
        "sift": v["sift_val"] is not None and v["sift_val"] < 0.05,
        "polyphen": pp.startswith(("probably_damaging", "possibly_damaging")),
        "conservation": v["conservation"] is not None and v["conservation"] >= 0.9,
    }

LETTER = {"cadd": "D", "sift": "S", "polyphen": "P", "conservation": "C"}  # D = CADD
for v in working_set:
    c = calls(v)
    v["n_damaging"] = sum(c.values())
    v["votes"] = "".join(LETTER[k] if val else "-" for k, val in c.items())

print("Votes by label:")
for label in ["Pathogenic", "Uncertain significance", "Likely benign", "Benign"]:
    cnt = Counter(int(v["n_damaging"]) for v in working_set if v["clinvar_label"] == label)
    print(f"  {label:24s}", dict(sorted(cnt.items())))

vus = [v for v in working_set if v["clinvar_label"] == "Uncertain significance"]
vus.sort(key=lambda v: -v["n_damaging"])
print("\nVUS, most damaging votes first  (votes = D S P C)")
for v in vus:
    print(f'{v["name"].split(":")[1]:34s} n={v["n_damaging"]} {v["votes"]}  '
          f'CADD={v["cadd_phred_val"]:5.1f} SIFT={v["sift_val"]:.2f} PP={v["polyphen_val"]:.2f} cons={v["conservation"]:.2f}')

pd.DataFrame(working_set)[cols + ["n_damaging", "votes"]].to_csv("kcnq1_classical_scores.csv", index=False)
```

```python
import requests, time, json, os

HALF = 3000
CACHE = "window_cache.json"
cache = json.load(open(CACHE)) if os.path.exists(CACHE) else {}

def fetch_ensembl(chrom, pos):
    url = f"https://rest.ensembl.org/sequence/region/human/{chrom}:{pos-HALF}..{pos+HALF}:1"
    r = requests.get(url, headers={"Content-Type": "text/plain"}, timeout=60)
    r.raise_for_status()
    return r.text.strip().upper()

def fetch_ucsc(chrom, pos):
    # UCSC uses 0-based start, end-exclusive, and "chr" prefixes
    url = (f"https://api.genome.ucsc.edu/getData/sequence?genome=hg38;"
           f"chrom=chr{chrom};start={pos-HALF-1};end={pos+HALF}")
    r = requests.get(url, timeout=60)
    r.raise_for_status()
    return r.json()["dna"].upper()

def fetch_window(chrom, pos, tries=4):
    for attempt in range(tries):
        try:
            return fetch_ensembl(chrom, pos)
        except requests.RequestException as e:
            wait = 5 * 2**attempt          # 5, 10, 20, 40 s
            print(f"  Ensembl failed ({e.__class__.__name__}); retry in {wait}s")
            time.sleep(wait)
    print("  Ensembl still failing, trying UCSC")
    return fetch_ucsc(chrom, pos)

rows, problems = [], []
for v in working_set:
    key = re.sub(r'\s*\(p\..*?\)\s*$', '', v["name"])
    g = vep_genomic[key]
    pos = int(g["pos"])
    ck = f"{g['chrom']}:{pos}"

    if ck not in cache:
        cache[ck] = fetch_window(g["chrom"], pos)
        json.dump(cache, open(CACHE, "w"))   # save progress after every variant
        time.sleep(0.3)
    ref_seq = cache[ck]

    if len(ref_seq) != 2*HALF + 1 or ref_seq[HALF] != g["ref"]:
        problems.append((v["name"], len(ref_seq), ref_seq[HALF] if len(ref_seq) > HALF else None, g["ref"]))
        continue

    alt_seq = ref_seq[:HALF] + g["alt"] + ref_seq[HALF+1:]
    rows.append({"name": v["name"], "clinvar_label": v["clinvar_label"],
                 "chrom": g["chrom"], "pos": pos, "ref": g["ref"], "alt": g["alt"],
                 "ref_seq": ref_seq, "alt_seq": alt_seq})

print(f"Windows built: {len(rows)} / {len(working_set)}")
print("Problems:", problems)
pd.DataFrame(rows).to_csv("kcnq1_lm_windows.csv", index=False)
```
