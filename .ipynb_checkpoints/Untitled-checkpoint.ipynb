```python
from Bio import AlignIO
import numpy as np

alignment = AlignIO.read("kcnq1_alligned.fasta", "fasta")
n_seqs = len(alignment)
aln_length = alignment.get_alignment_length()

def shannon_entropy(column):
    """
    column = string of bases at one alignment position, one per species.
    Lower entropy = more conserved. 0 = all species identical.
    """
    bases = [b.upper() for b in column if b != '-']  # ignore gaps
    if len(bases) == 0:
        return None  # entirely gaps at this position, skip
    counts = {}
    for b in bases:
        counts[b] = counts.get(b, 0) + 1
    total = len(bases)
    entropy = 0
    for count in counts.values():
        p = count / total
        entropy -= p * np.log2(p)
    return entropy

conservation_scores = []
for i in range(aln_length):
    column = alignment[:, i]  # all species' base at position i
    entropy = shannon_entropy(column)
    conservation_scores.append(entropy)

# convert entropy to a conservation score where HIGHER = MORE conserved
# (entropy is the opposite: 0 entropy = fully conserved)
max_entropy = np.log2(4)  # max possible for 4 DNA bases
conservation = [
    (max_entropy - e) / max_entropy if e is not None else None
    for e in conservation_scores
]
```

```python
def load_and_parse(filepath):
    variants = []
    with open(filepath) as f:
        reader = csv.DictReader(f, delimiter='\t')
        for row in reader:
            name = row["Name"]
            clinical_sig = row["Germline classification"]
            parsed = parse_hgvs_cds(name)
            if parsed is None:
                continue
            cds_pos, ref, alt = parsed
            variants.append({
                "name": name,
                "cds_pos": cds_pos,
                "ref": ref,
                "alt": alt,
                "clinvar_label": clinical_sig,  # actual value from the file, not assumed
            })
    return variants

vus_all = load_and_parse("clinvar_download_20260903_120920.tsv")
pathogenic_all = load_and_parse("clinvar_download_20260903_120920.tsv")
benign_all = load_and_parse("clinvar_download_20260903_120920.tsv")

print(f"Available: {len(vus_all)} VUS, {len(pathogenic_all)} pathogenic, {len(benign_all)} benign")

# check what labels actually showed up — catches combined categories
print("Pathogenic file labels:", set(v["clinvar_label"] for v in pathogenic_all))
print("Benign file labels:", set(v["clinvar_label"] for v in benign_all))
```

```python
import csv
import re

def parse_hgvs_cds(name_string):
    match = re.search(r'c\.(\d+)([ACGT])>([ACGT])', name_string)
    if match:
        pos, ref, alt = match.groups()
        return int(pos), ref, alt
    return None

variants = []
with open("clinvar_download_20260903_120920.tsv") as f:
    reader = csv.DictReader(f, delimiter='\t')
    for row in reader:
        if row.get("Gene(s)", "").strip() != "KCNQ1":
            continue
        label = row["Germline classification"].strip()
        if label not in ("Pathogenic", "Benign", "Uncertain significance"):
            continue
        parsed = parse_hgvs_cds(row["Name"])
        if parsed is None:
            continue
        cds_pos, ref, alt = parsed
        variants.append({
            "name": row["Name"],
            "cds_pos": cds_pos,
            "ref": ref,
            "alt": alt,
            "clinvar_label": label,
        })

print(f"Kept {len(variants)} variants after gene/label/parse filtering")
```

```python
def build_position_map(alignment, human_seq_id):
    """
    Returns a dict: CDS position (1-based, ungapped) -> alignment column index (0-based)
    """
    human_row = None
    for record in alignment:
        if human_seq_id in record.id:
            human_row = str(record.seq)
            break
    if human_row is None:
        raise ValueError(f"Couldn't find {human_seq_id} in alignment")

    position_map = {}
    cds_pos = 0
    for col_idx, base in enumerate(human_row):
        if base != '-':
            cds_pos += 1
            position_map[cds_pos] = col_idx
    return position_map

position_map = build_position_map(alignment, "NM_000218.3:92-2122")
```

```python
variants = [
    {"id": "example1", "cds_pos": 1032, "clinvar_label": "Uncertain significance"},
    # ... fill in your actual pulled variants here
]

for v in variants:
    col = position_map.get(v["cds_pos"])
    if col is None:
        print(f"{v['id']}: position {v['cds_pos']} not found in map — check numbering")
        continue
    score = conservation[col]
    v["conservation_score"] = score
    print(f"{v['id']} (CDS {v['cds_pos']}, {v['clinvar_label']}): conservation = {score}")
```

```python
import random
random.seed(42)

# split your already-parsed 1600 variants by their real label
pathogenic_all = [v for v in variants if v["clinvar_label"] == "Pathogenic"]
benign_all = [v for v in variants if v["clinvar_label"] == "Benign"]
vus_all = [v for v in variants if v["clinvar_label"] == "Uncertain significance"]

print(f"Available: {len(vus_all)} VUS, {len(pathogenic_all)} pathogenic, {len(benign_all)} benign")

vus_sample = random.sample(vus_all, min(25, len(vus_all)))
pathogenic_sample = random.sample(pathogenic_all, min(15, len(pathogenic_all)))
benign_sample = random.sample(benign_all, min(15, len(benign_all)))

working_set = vus_sample + pathogenic_sample + benign_sample
print(f"Working set: {len(working_set)} total variants")
```

```python
for v in working_set:
    col = position_map.get(v["cds_pos"])
    if col is None:
        print(f"{v['name']}: position {v['cds_pos']} not found in map — check numbering")
        continue
    v["conservation_score"] = conservation[col]

# quick look at results, grouped by label
for label in ["Pathogenic", "Benign", "Uncertain significance"]:
    scores = [v["conservation_score"] for v in working_set
              if v["clinvar_label"] == label and "conservation_score" in v]
    if scores:
        print(f"{label}: n={len(scores)}, mean conservation={sum(scores)/len(scores):.3f}")
```
